# immune checkpoint block-ade (ICB)

> immune checkpoint block-ade (ICB)

**Immune checkpoint blockade (ICB)**, also called **immune checkpoint inhibitor therapy**, is a type of cancer immunotherapy.

It works by blocking inhibitory “checkpoint” signals that tumors can exploit to suppress T-cell activity. Removing these brakes can help the immune system recognize and attack cancer cells.

Common ICB targets include:

- **PD-1**: inhibitors include pembrolizumab and nivolumab
- **PD-L1**: inhibitors include atezolizumab, durvalumab, and avelumab
- **CTLA-4**: inhibitor includes ipilimumab

In Chinese, the usual translation is **免疫检查点阻断** or, more commonly in clinical writing, **免疫检查点抑制剂治疗**.

For paper writing, use:

> immune checkpoint blockade (ICB)

Avoid splitting the word as `block-ade`; it should be written as one word: `blockade`.

**第 1 周：肿瘤免疫的最小核心框架**

先把这些概念连成一条过程，而不是孤立背术语：

1. 肿瘤抗原被树突状细胞（DC）摄取。
2. DC 成熟并将抗原呈递给初始 T 细胞。
3. CD8+ 细胞毒性 T 细胞（CTL）扩增、进入肿瘤并杀伤肿瘤细胞。
4. 肿瘤通过 PD-L1/PD-1、CTLA-4、Treg、M2 型 TAM 和 MDSC 压制这一过程。
5. ICB 解除部分抑制，但前提是肿瘤中已经有可被重新激活的免疫反应。
6. ICD 的价值在于补上“抗原释放、DC 激活和 T 细胞浸润”这一步。

这一周重点掌握的词：

`APC, DC, MHC-I, MHC-II, CD8+ T cell, CTL, CD4+ T cell, Treg, macrophage, M1/M2, TAM, MDSC, PD-1, PD-L1, CTLA-4, ICB, cold tumor, hot tumor`

你之前问的 `immune checkpoint blockade (ICB)` 就在这里：它不是直接杀伤肿瘤，而是解除 T 细胞被 PD-1/PD-L1 或 CTLA-4 等通路抑制的状态。

**第 2 周：ICD 与细胞死亡**

这是这篇综述最关键的“免疫学和细胞生物学交叉点”。需要特别弄清：

- **普通细胞死亡**：可能只造成炎症，甚至诱导免疫耐受。
- **ICD**：不仅有肿瘤抗原，还提供危险信号，使免疫系统把死亡细胞识别为“需要处理并建立适应性免疫”的对象。
- **DAMPs**：
  - `CRT` 外翻，常被称为“eat-me signal”，促进吞噬；
  - `ATP` 外排，募集并激活免疫细胞；
  - `HMGB1` 释放，参与 DC 激活和抗原呈递；
  - `HSP70/HSP90` 暴露，帮助免疫识别和抗原处理。
- **ROS - ER stress - DAMPs 暴露**：这是本文把 SDT 和免疫治疗连起来的关键机制。
- **细胞死亡类型**：先了解凋亡、坏死性凋亡、焦亡和铁死亡的定义、典型分子标志物及免疫后果即可。不要一开始陷入每条信号通路的细节。

建议你对每一个 ICD 相关图都问同一组问题：

1. 细胞是怎么死的？
2. 作者证明它是 ICD 而非一般细胞毒性了吗？
3. 检测了哪些 DAMPs？
4. DC 是否成熟？
5. CD8+ T 细胞是否增加或激活？
6. 是否证明了免疫记忆、远端肿瘤抑制或抗转移效果？

**第 3 周：声动力治疗的物理化学基础**

本文不是传统药理或纯免疫论文，因此需要补一层“超声如何变成生物效应”。

要掌握：

- 超声基本量：频率、声强、占空比、治疗时间。
- **声空化（acoustic cavitation）**：超声导致微泡振荡甚至坍塌，可产生局部高温高压、冲击波、剪切力和声致发光。
- **声敏剂（sonosensitizer）**：在超声激发下参与 ROS 生成的分子或材料。
- ROS 的主要类型：`1O2`、`O2.-`、`.OH`、`H2O2`；它们的来源、寿命、扩散范围和生物学后果。
- SDT 与 PDT/RDT 的比较：
  - PDT：光激发，空间可控但组织穿透较浅；
  - RDT：X 射线穿透深，但电离辐射毒性是问题；
  - SDT：超声穿透深，局部可聚焦，但依赖声敏剂、氧供应和声学参数。

读论文时必须特别看清：作者的 ROS 证据是细胞内探针、ESR、自旋捕获、单线态氧检测，还是只凭材料机理推测。它们证明力并不相同。

**第 4 周：肿瘤微环境与纳米平台**

这篇文章中的多数“复杂纳米材料设计”都在解决 SDT 的几个实际瓶颈：

| SDT/免疫治疗的瓶颈 | 常见设计策略 | 你应理解的机制 |
|---|---|---|
| 肿瘤缺氧，ROS 产量低 | 供氧、催化产氧、降低氧耗 | 氧是多数 ROS 生成路径的重要底物 |
| 肿瘤内 GSH 高，清除 ROS | GSH 消耗或氧化还原响应设计 | GSH 会削弱氧化治疗的有效性 |
| 声敏剂到不了肿瘤或进不了细胞 | 被动靶向、主动靶向、细胞器靶向 | 递送位置决定 ROS 是否能造成有效损伤 |
| ICD 后仍有免疫抑制 | 联用 ICB、TLR 激动剂、STING 激动剂或巨噬细胞重编程 | 免疫启动后仍需解除抑制并改善 TME |
| 原发瘤控制但转移/复发存在 | 全身性免疫激活、免疫记忆、联合疗法 | 局部物理治疗需要转化为系统性抗肿瘤免疫 |

这里你不必先系统学完所有纳米材料。先建立一个固定阅读模板：

```text
材料由什么组成？
每个组分分别承担什么任务？
它如何到达或滞留于肿瘤？
超声后发生什么物理/化学变化？
ROS 或细胞死亡怎样被增强？
怎样证明产生 ICD？
怎样证明最终是免疫效应而非单纯局部杀伤？
毒性、体内分布和可转化性是否被评估？
```

**后续再补：实验方法与数据判读**

当你开始读原始研究论文时，以下方法会频繁出现，建议按“它回答什么问题”来学：

- `CCK-8/MTT、Annexin V/PI`：细胞是否死亡、死亡类型大致如何。
- `DCFH-DA、SOSG、ESR`：是否产生 ROS、产生何种 ROS。
- `CRT 暴露、ATP 释放、HMGB1 释放`：是否存在 ICD 相关证据。
- `CD80/CD86`：DC 是否成熟。
- `CD3、CD4、CD8、Foxp3、F4/80、CD206`：肿瘤浸润 T 细胞、Treg 和巨噬细胞状态。
- `流式细胞术`：不同免疫细胞群体的比例和表型。
- `肿瘤生长曲线、存活曲线、转移模型、再攻击模型`：局部疗效、系统免疫和免疫记忆。
- `H&E、TUNEL、免疫荧光、IHC`：组织层面的损伤、细胞死亡和免疫浸润。
